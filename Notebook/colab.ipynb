```python
# Clean text values by removing commas, brackets, and mapping dashes ('–') to '0'
df['Value_Cr'] = (
    df['Raw_Value_as_reported_TEXT']
    .str.replace(',', '', regex=False)
    .str.replace('(', '-', regex=False)
    .str.replace(')', '', regex=False)
    .str.replace('–', '0', regex=False)   # Handle en-dash
    .str.replace('—', '0', regex=False)   # Handle em-dash
    .str.strip()
    .astype(float)
)

df[['Raw_Value_as_reported_TEXT', 'Value_Cr']].head(15)
```

```python
revenue = (
    df[df['Line_Item_as_reported'] == 'Revenue from operations']
    .set_index('Fiscal_Year')['Value_Cr']
    .sort_index()
)

revenue_growth = revenue.pct_change() * 100

revenue_analysis = pd.DataFrame({
    'Revenue_Cr': revenue,
    'Revenue_Growth_%': revenue_growth
})

revenue_analysis.round(2)
```
